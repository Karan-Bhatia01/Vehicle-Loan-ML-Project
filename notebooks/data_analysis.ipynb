### Data Cleaning

```python
df['Employment.Type'].value_counts()
```

```python
df['Employment.Type'] = df['Employment.Type'].fillna('Salaried')
```

```python
df['Employment.Type'].value_counts()
```

```python
df['Date.of.Birth'] = pd.to_datetime(df['Date.of.Birth'], format='%d-%m-%y')
df['DisbursalDate'] = pd.to_datetime(df['DisbursalDate'], format='%d-%m-%y')
```

```python
df['Date.of.Birth'] = df['Date.of.Birth'].apply(lambda x: x - pd.DateOffset(years=100) if x.year > 2026 else x)
```

```python
def convert_to_months(duration_str):
    
    if pd.isna(duration_str):
        return 0
    # regex to extract the digits before 'yrs' and 'mon'
    match = re.search(r'(\d+)yrs (\d+)mon', str(duration_str))
    if match:
        years = int(match.group(1))
        months = int(match.group(2))
        return (years * 12) + months
    return 0
```

```python
df['AVERAGE.ACCT.AGE'] = df['AVERAGE.ACCT.AGE'].apply(convert_to_months)
df['CREDIT.HISTORY.LENGTH'] = df['CREDIT.HISTORY.LENGTH'].apply(convert_to_months)
```

```python
text_cols = ['Employment.Type', 'PERFORM_CNS.SCORE.DESCRIPTION']
for col in text_cols:
    df[col] = df[col].astype('category')
```

```python
# Memory Optimization: Downcast binary flags to int8
flag_cols = [
    'MobileNo_Avl_Flag', 'Aadhar_flag', 'PAN_flag', 
    'VoterID_flag', 'Driving_flag', 'Passport_flag', 'loan_default'
]
for col in flag_cols:
    df[col] = df[col].astype('int8')
```

```python
df.sample(5)
```

```python
# Drop columns with zero variance (only 1 unique value)
df = df.drop(columns=['MobileNo_Avl_Flag'])
```

```python
# Check if disbursed amount is greater than asset cost (a potential data error)
anomalies = df[df['disbursed_amount'] > df['asset_cost']]
print(f"Rows where disbursed amount > asset cost: {len(anomalies)}")
```

```python
import datetime
# If a birth year is greater than the current year, subtract 100 years
df['Date.of.Birth'] = df['Date.of.Birth'].apply(
    lambda x: x - pd.DateOffset(years=100) if x.year > 2026 else x
)
# Now it is safe to calculate Age at Disbursal
df['Age_at_Disbursal'] = (df['DisbursalDate'] - df['Date.of.Birth']).dt.days // 365
# Check if there are any underage/impossible ages (e.g., < 18 or > 100)
print(df['Age_at_Disbursal'].describe())
```

### Feature Engineering

```python
# Total Account Counts
df['TOTAL.NO.OF.ACCTS'] = df['PRI.NO.OF.ACCTS'] + df['SEC.NO.OF.ACCTS']
df['TOTAL.ACTIVE.ACCTS'] = df['PRI.ACTIVE.ACCTS'] + df['SEC.ACTIVE.ACCTS']
df['TOTAL.OVERDUE.ACCTS'] = df['PRI.OVERDUE.ACCTS'] + df['SEC.OVERDUE.ACCTS']

# Total Financial Exposure
df['TOTAL.CURRENT.BALANCE'] = df['PRI.CURRENT.BALANCE'] + df['SEC.CURRENT.BALANCE']
df['TOTAL.SANCTIONED.AMOUNT'] = df['PRI.SANCTIONED.AMOUNT'] + df['SEC.SANCTIONED.AMOUNT']
df['TOTAL.INSTAL.AMT'] = df['PRIMARY.INSTAL.AMT'] + df['SEC.INSTAL.AMT']
```

```python
# 1. Overdue Ratio: Proportion of accounts that are overdue
# We add +1 to the denominator to prevent Division By Zero errors
df['OVERDUE_RATIO'] = df['TOTAL.OVERDUE.ACCTS'] / (df['TOTAL.NO.OF.ACCTS'] + 1)

# 2. Credit Utilization Ratio: How much of their sanctioned loan amount are they using?
# High utilization usually indicates financial stress
df['CREDIT_UTILIZATION'] = df['TOTAL.CURRENT.BALANCE'] / (df['TOTAL.SANCTIONED.AMOUNT'] + 1)

# 3. Active Account Ratio: Proportion of accounts still active
df['ACTIVE_RATIO'] = df['TOTAL.ACTIVE.ACCTS'] / (df['TOTAL.NO.OF.ACCTS'] + 1)
```

```python
# How much upfront cash did the customer pay?
df['DOWNPAYMENT'] = df['asset_cost'] - df['disbursed_amount']

# Ratio of downpayment to the total asset cost
df['DOWNPAYMENT_RATIO'] = df['DOWNPAYMENT'] / df['asset_cost']
```

```python
# Sum of all ID flags provided
df['TOTAL_IDS_PROVIDED'] = (
    df['Aadhar_flag'] + df['PAN_flag'] + 
    df['VoterID_flag'] + df['Driving_flag'] + 
    df['Passport_flag']
)
```

```python
cols_to_drop = [
    # Identifiers and Dates
    'UniqueID', 'DisbursalDate', 'MobileNo_Avl_Flag',
    
    # Redundant Text Column
    'PERFORM_CNS.SCORE.DESCRIPTION',
    
    # Redundant Secondary Account Columns (since we have TOTAL)
    'SEC.NO.OF.ACCTS', 'SEC.ACTIVE.ACCTS', 'SEC.OVERDUE.ACCTS', 
    'SEC.CURRENT.BALANCE', 'SEC.SANCTIONED.AMOUNT', 'SEC.DISBURSED.AMOUNT', 'SEC.INSTAL.AMT'
]

# Only drop columns that currently exist in the dataframe to avoid errors
existing_cols_to_drop = [col for col in cols_to_drop if col in df.columns]

df = df.drop(columns=existing_cols_to_drop)

print(f"Dropped {len(existing_cols_to_drop)} columns. New shape: {df.shape}")
```

```python
# Assuming you converted these to total months earlier
# A large delta means a flurry of recent new accounts
df['CREDIT_HUNGER_DELTA'] = df['CREDIT.HISTORY.LENGTH'] - df['AVERAGE.ACCT.AGE']

# You can also look at the ratio
df['CREDIT_HUNGER_RATIO'] = df['AVERAGE.ACCT.AGE'] / (df['CREDIT.HISTORY.LENGTH'] + 1)
```

```python
# 1. Does the borrower have ANY credit history? 
# (Score of 0 or 'No Bureau History Available' usually means no history)
df['HAS_CREDIT_HISTORY'] = (df['PERFORM_CNS.SCORE'] > 10).astype(int)

# 2. Does the borrower have ANY overdue accounts right now?
df['HAS_OVERDUE_ACCTS'] = (df['TOTAL.OVERDUE.ACCTS'] > 0).astype(int)

# 3. Did they open ANY new accounts in the last 6 months?
df['RECENT_NEW_ACCT'] = (df['NEW.ACCTS.IN.LAST.SIX.MONTHS'] > 0).astype(int)

# 4. Young Borrower Flag (e.g., under 25)
df['IS_YOUNG_BORROWER'] = (df['Age_at_Disbursal'] < 25).astype(int)
```

```python
cols_to_drop = [
    'UniqueID',                      # Just an index
    'Date.of.Birth',                 # Replaced by Age_at_Disbursal
    'DisbursalDate',                 # Replaced by Age_at_Disbursal
    'MobileNo_Avl_Flag',             # Every row is exactly the same (1)
    'PERFORM_CNS.SCORE.DESCRIPTION'  # Text version of PERFORM_CNS.SCORE
]
# If you combined Primary and Secondary accounts earlier, you can also add all the 'SEC.' columns to this list.

existing_to_drop = [c for c in cols_to_drop if c in df.columns]
df = df.drop(columns=existing_to_drop)
```

### Data Cleaning & Feature Engineering Summary

### 1. Data Cleaning & Optimization

- **Missing Value Imputation**
  - Filled **7,661 missing values** in `Employment.Type` using the mode (`Salaried`) to preserve dataset completeness.

- **Date Parsing & Correction**
  - Converted `Date.of.Birth` and `DisbursalDate` into proper datetime objects.
  - Corrected the **100-year parsing bug** in birth dates.
  - Made an accurate `Age_at_Disbursal` feature with values ranging from **18–69 years**.

- **Duration Parsing**
  - Parsed inconsistent duration strings (e.g., `"1yrs 11mon"`) into total months.
  - Applied this conversion to:
    - `AVERAGE.ACCT.AGE`
    - `CREDIT.HISTORY.LENGTH`

- **Memory Optimization**
  - Downcasted binary columns to `int8`.
  - Converted categorical text columns to the `category` data type.


### 2. Advanced Feature Engineering

#### Total Financial Exposure
Combined Primary (`PRI`) and Secondary (`SEC`) bureau accounts to generate consolidated financial indicators:

- `TOTAL.NO.OF.ACCTS`
- `TOTAL.CURRENT.BALANCE`
- `TOTAL.SANCTIONED.AMOUNT`
- `TOTAL.DISBURSED.AMOUNT`
- Other aggregate credit exposure metrics

#### Risk Ratios

Created predictive ratio-based features to quantify financial health:

- **`OVERDUE_RATIO`** – Measures the proportion of overdue payments relative to outstanding balances.
- **`CREDIT_UTILIZATION`** – Captures how much sanctioned credit is currently utilized.
- **`DOWNPAYMENT_RATIO`** – Represents the borrower's own financial contribution toward the asset, indicating commitment ("skin in the game").

#### Credit Behavior

Engineered **`CREDIT_HUNGER_DELTA`**, defined as:

> Total Credit History − Average Account Age

This feature helps identify borrowers who have opened several new credit accounts within a short period, often indicating aggressive borrowing behavior.

#### Behavioral Flags

Generated boolean indicators to provide tree-based models with clean decision boundaries:

- `HAS_CREDIT_HISTORY`
- `HAS_OVERDUE_ACCTS`
- `IS_YOUNG_BORROWER`

### 3. Dimensionality Reduction

- **Raw dates and identifiers**
  - `Date.of.Birth`
  - `DisbursalDate`
  - `UniqueID`

- **Constant Feature**
  - `MobileNo_Avl_Flag` (zero variance)

- **Redundant Features**
  - Original text-based duration columns after conversion
  - All `SEC.*` account columns, as their information was fully incorporated into the  `TOTAL.*` features

This reduced dataset complexity while preserving all relevant predictive information for downstream machine learning models.
